```python
from pyspark.sql.functions import * 
from pyspark.sql.types import *
from pyspark.sql.window import Window
from pyspark.sql import functions as F
```

```python
df = spark.read.table('fintech360.schema_validation.bureau_data')
display(df)
```

```python
df = df.withColumn("report_ts", unix_timestamp(col("report_ts"))) 
```

```python
df.display()
```

```python
df = df.withColumn(
    "delinquency_ever",
    when(col("delinquency_ever") == True, 1).otherwise(0)
)

display(df)
```

```python
df = df.withColumn(
    "score_band",
    regexp_extract(
        col("bureau_details_json"),
        r'score_band"":""([^"]+)',
        1
    )
)

display(df.select("bureau_details_json", "score_band"))
```

```python
df.display()
```

```python
from pyspark.sql import functions as F

df = (
    df
    .withColumn("created_at", F.col("created_at").cast("long") * 1000)
    .withColumn("updated_at", F.col("updated_at").cast("long") * 1000)
)

df.select("created_at", "updated_at").show(10, truncate=False)
```

```python
df = df.withColumn("_ingested_at", unix_timestamp(col("_ingested_at").cast("timestamp")))
df.limit(5).show()
```

```python
(df.write
 .format("delta")
 .mode('append')
 .saveAsTable('fintech360.silver.bureau_data'))
```
