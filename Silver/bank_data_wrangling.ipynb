```python
from pyspark.sql.functions import * 
from pyspark.sql.types import *
from pyspark.sql.window import Window
from pyspark.sql import functions as F
```

```python
df = spark.read.table('fintech360.schema_validation.bank_data')
display(df)
```

# timestamp to epoch

```python
df = df.withColumn("transaction_ts", unix_timestamp(col("transaction_ts")))
display(df) 
```

# is_salary_like text to 0/1

```python
df = df.withColumn("is_salary_like",when(col("is_salary_like") == "true",1).otherwise(0))
```

```python
display(df)
```

```python
from pyspark.sql import functions as F

df = (
    df
    .withColumn("created_at", F.col("created_at").cast("long") * 1000)
    .withColumn("updated_at", F.col("updated_at").cast("long") * 1000)
)

df.select("created_at", "updated_at").show(10, truncate=False)
```

```python
df = df.withColumn("_ingested_at", unix_timestamp(col("_ingested_at").cast("timestamp")))
df.limit(5).show()
```

```python
(df.write
 .format("delta")
 .mode("append")
 .saveAsTable('fintech360.silver.bank_data'))
```
